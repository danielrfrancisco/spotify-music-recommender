# **1. Exploración inicial del dataset de Spotify**

**Objetivo:** Inspeccionar la estructura del conjunto de datos antes de realizar
cualquier transformación o limpieza.

**Fuente de datos:** Dataset de Kaggle: Spotify Tracks Dataset - Audio Features.

**Archivo:** `spotify-tracks-dataset-detailed.csv`

**Procedimiento:**
1. Cargar el archivo CSV.
2. Verificar que la carga se haya realizado correctamente.
3. Examinar las dimensiones, columnas y tipos de datos.
4. Identificar posibles problemas que requieran un análisis posterior.

**Criterio de trabajo:** En esta etapa inicial no se modificarán ni eliminarán datos. Primero se observará su estructura para fundamentar las decisiones de limpieza posteriores.


```python
import pandas as pd
```

```python
from pathlib import Path

print('Directorio de trabajo:', Path.cwd())
```

```python
ruta_datos = Path.cwd().parent / 'data' / 'spotify-tracks-dataset-detailed.csv'

print('Ruta del archivo:', ruta_datos)
print('¿Existe el archivo?', ruta_datos.is_file())
```

```python
df = pd.read_csv(ruta_datos)

print('Archivo cargado correctamente.')
print('Tipo de objeto:', type(df))
print('Dimensiones:', df.shape)
```

```python
duplicados_completos = df.duplicated()

print('Filas completamente duplicadas:', duplicados_completos.sum())
```

```python

duplicados_id = df['track_id'].duplicated()

print('Filas con track_id repetido:', duplicados_id.sum())
print('Identificadores únicos:', df['track_id'].nunique())
print('Filas totales:', len(df))
```

```python
frecuencia_id = df['track_id'].value_counts()

print('Identificadores que aparecen más de una vez:', (frecuencia_id > 1).sum())

print('Máximo de apariciones de un identificador:', frecuencia_id.max())

print('\nDistribución de las frecuencias:')
print(frecuencia_id.value_counts().sort_index().head(10))
```

```python
ids_repetidos = frecuencia_id[frecuencia_id > 1].index

df.loc[
    df['track_id'].isin(ids_repetidos),
    ['track_id', 'artists', 'album_name', 'track_name', 'track_genre']
].sort_values(
    ['track_id']
).head(15)
```

```python
columnas_comparacion = [
    columna for columna in df.columns
    if columna not in ['track_id']
]

variacion_por_id = (
    df[df['track_id'].duplicated(keep = False)]
    .groupby('track_id')[columnas_comparacion]
    .nunique(dropna = False)
)

print('¿Hay diferencias en las demás columnas?')
print(variacion_por_id.gt(1).any().any())

print('\nColumnas con diferencias:')
print(variacion_por_id.columns[variacion_por_id.gt(1).any()])
```

```python
generos_por_id = (
    df[df['track_id'].duplicated(keep = False)]
    .groupby('track_id')['track_genre']
    .nunique()
)

print('IDs con más de un género:', (generos_por_id > 1).sum())

print('\nDistribución de géneros distintos por ID:')
print(generos_por_id.value_counts().sort_index())
```

```python
popularidad_por_id = (
    df[df['track_id'].duplicated(keep = False)]
    .groupby('track_id')['popularity']
    .nunique()
)

print('IDs con más de un valor de popularidad:', (popularidad_por_id > 1).sum())

print('\nDistribución de valores distintos de popularidad:')
print(popularidad_por_id.value_counts().sort_index())
```

```python
ids_popularidad_variable = popularidad_por_id[popularidad_por_id > 1].index

df.loc[
    df['track_id'].isin(ids_popularidad_variable),
    ['track_id', 'artists', 'track_name', 'popularity', 'track_genre']
].sort_values('track_id').head(20)
```

```python
popularidad_por_id = (
    df[df['track_id'].duplicated(keep = False)]
    .groupby('track_id')['popularity']
    .agg(['min', 'max', 'nunique'])
)

popularidad_por_id['diferencia'] = (popularidad_por_id['max'] - popularidad_por_id['min'])

ids_popularidad_variable = popularidad_por_id[popularidad_por_id['nunique'] > 1]

print('Número de IDs con popularidad variable:', len(ids_popularidad_variable))

print('\nDistribución de las diferencias de popularidad:')
print(ids_popularidad_variable['diferencia'].value_counts().sort_index())

print('\nResumen estadístico de las diferencias:')
print(ids_popularidad_variable['diferencia'].describe())
```

```python
columnas_numericas = df.select_dtypes(include = 'number').columns

df[columnas_numericas].describe().T
```

```python
print('Valores sospechosos por variable:\n')

print('Duración igual o menor que cero:', (df['duration_ms'] <= 0).sum())
print('Duración superior a 10 minutos:', (df['duration_ms'] > 600000).sum())

print('Loudness positivo:', (df['loudness'] > 0).sum())

print('Tempo igual o menor que cero:', (df['tempo'] <= 0).sum())

print('Time signature igual a cero:', (df['time_signature'] == 0).sum())
print('Time signature igual a cinco:', (df['time_signature'] == 5).sum())
```

```python
df.loc[
    df['loudness'] > 0,
    ['track_id', 'artists', 'track_name', 'loudness', 'track_genre']
].sort_values('loudness', ascending = False).head(10)
```

```python
df.loc[
    df['tempo'] <= 0,
    ['track_id', 'artists', 'track_name', 'tempo', 'track_genre']
].sort_values('tempo').head(10)
```

```python
df.loc[
    df['time_signature'] == 0,
    ['track_id', 'artists', 'track_name', 'time_signature', 'track_genre']
].head(10)
```

```python
print(
    'Tempo y time signature iguales a cero:',
    ((df['tempo'] == 0) & (df['time_signature'] == 0)).sum()
)
```

```python
duplicados_exactos = df[df.duplicated(keep = False)]

print('Filas que pertenecen a grupos de duplicados exactos:',
      len(duplicados_exactos))

print('\nIDs únicos entre esas filas:',
      duplicados_exactos['track_id'].nunique())

print('\nGéneros presentes entre esas filas:')
print(duplicados_exactos['track_genre'].value_counts())
```

```python
tamanos_grupos = (
    duplicados_exactos
    .groupby(list(df.columns), dropna = False)
    .size()
)

print('Distribución del tamaño de los grupos duplicados:')
print(tamanos_grupos.value_counts().sort_index())

print('\nNúmero de grupos de duplicados exactos:',
      len(tamanos_grupos))
```
